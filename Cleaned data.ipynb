## Key Findings

1. **Overall Churn:** Approximately 23.87% of customers with a recorded churn status have churned.

2. **Subscription Type:** Basic plan customers have the lowest churn rate (17.6%), while Standard plan customers have the highest (28.4%).

3. **Contract Type:** Month-to-Month customers have a churn rate of 21.1%, compared with 25.3% for One Year and 25.9% for Two Year contracts.

4. **Customer Tenure:** Customers who churned had a lower average tenure (33.33 months) than customers who stayed (37.49 months).

5. **Payment Method:** Debit Card users have a churn rate of 28.4%, while Cash users have the lowest rate at 17.5%.

6. **Internet Service:** Cable users have a churn rate of 28.8%, the highest among the recorded internet service categories.

7. **Data Quality:** Duplicate records were removed, inconsistent categorical values were standardized, invalid numerical values were converted to missing values, and dates were standardized. Existing missing values were retained rather than filled.


## Conclusion

This project focused on cleaning and exploring an e-commerce customer churn dataset using Python and Pandas. The analysis examined customer churn patterns across subscription types, contract types, payment methods, internet services, and customer tenure.

The findings show differences in churn rates across customer groups. Standard plan and Cable internet users had comparatively higher churn rates, while Basic plan and Cash payment users had lower churn rates. Customers who churned also had a lower average tenure than customers who remained.

These findings can help a business identify customer segments that may need further investigation and develop targeted customer-retention strategies. However, the observed relationships do not establish causation. Further analysis would be required to understand the underlying reasons for customer churn.

Overall, the project provided practical experience in data cleaning, exploratory data analysis, visualization, and communicating data-driven insights using Python.
